Explicit  we can do it in 2 places 
1.  at the time of Create array Element

```python
arr=np.array([10,20,30])
arr.dtype
```

```python
# type 1  using dtype
arr=np.array([10,20,30],dtype=np.int8)
print(arr)
print(arr.dtype)
```

```python
# type 2 using astype
a=[2,4,6,8]
a=np.array(a).astype('int8')
a.dtype

```

```python
a=[120,300,1,2,3]
a=np.array(a,dtype=np.int8)
print(a)
print(a.dtype)

# - 128 0 127
```

```python
a=[-129,10,1,2,3]
a=np.array(a,dtype=np.int8)
print(a)
print(a.dtype)
```

```python
a=[120,301,1,2,3]
a=np.array(a).astype('int8')
a  # 301-256
```
