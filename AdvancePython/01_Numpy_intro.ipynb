```python
import time
a=[1,2,3,4]
start_time=time.time() 
print([i*2 for i in a ])
end_time=time.time() 

print("Time Taken ",end_time-start_time)
```

```python
a=[1,2,3,4]
start_time=time.time() 
a=np.array(a)
print(a*2)
end_time=time.time() 

print("Time Taken ",end_time-start_time)
```
